```python
# Entity
df["student_id"] = df["Student_ID"].astype(str)

# Derived feature 1
df["cgpa_percent"] = (
    df["CGPA"] * 10
).astype("float32")

# Derived feature 2
df["skill_readiness_score"] = (
    df[
        [
            "cgpa_percent",
            "Programming_Skill_Score",
            "Communication_Skill_Score",
            "Soft_Skills_Score",
            "Curriculum_Coverage_Percent",
            "Industry_Demand_Score",
            "Curriculum_Industry_Alignment_Score"
        ]
    ].mean(axis=1)
).astype("float32")
```

```python
df["specialization_encoded"] = (
    df["Specialization"]
    .astype("category")
    .cat.codes
    .astype("int64")
)
```

```python
df.head()
```

```python
base_time = pd.Timestamp(
    "2026-01-01",
    tz="UTC"
)

df["event_timestamp"] = (
    base_time +
    pd.to_timedelta(
        range(len(df)),
        unit="s"
    )
)

df["created_timestamp"] = (
    df["event_timestamp"] +
    pd.Timedelta(seconds=1)
)
```

```python
feature_df.to_parquet(
    f"{repo_path}/data/student_features.parquet",
    index=False
)
```

```python
feature_store_yaml = """
project: student_skill_project

registry: data/registry.db

provider: local

offline_store:
  type: file

online_store:
  type: sqlite
  path: data/online_store.db
"""

with open(
    f"{repo_path}/feature_store.yaml",
    "w"
) as f:
    f.write(feature_store_yaml)
```

```python
feature_definition = '''
from datetime import timedelta

from feast import (
    Entity,
    FeatureView,
    FeatureService,
    Field,
    FileSource
)

from feast.types import (
    Float32,
    Int64
)


# -----------------------------
# ENTITY
# -----------------------------

student = Entity(
    name="student",
    join_keys=["student_id"],
    description="CSE student"
)


# -----------------------------
# DATA SOURCE
# -----------------------------

student_source = FileSource(
    name="student_source",
    path="data/student_features.parquet",
    timestamp_field="event_timestamp",
    created_timestamp_column="created_timestamp"
)


# -----------------------------
# FEATURE VIEW
# -----------------------------

student_feature_view = FeatureView(
    name="student_skill_features",
    entities=[student],

    ttl=timedelta(days=36500),

    schema=[
        Field(name="cgpa_percent", dtype=Float32),
        Field(name="Programming_Skill_Score", dtype=Int64),
        Field(name="Communication_Skill_Score", dtype=Int64),
        Field(name="Certifications_Count", dtype=Int64),
        Field(name="Internship_Experience", dtype=Int64),
        Field(name="Projects_Completed", dtype=Int64),
        Field(name="DSA_Platform_Rating", dtype=Int64),
        Field(name="Placement_Training_Attended", dtype=Int64),
        Field(name="Soft_Skills_Score", dtype=Int64),
        Field(name="Curriculum_Coverage_Percent", dtype=Int64),
        Field(name="Industry_Demand_Score", dtype=Int64),
        Field(name="Curriculum_Industry_Alignment_Score", dtype=Int64),
        Field(name="specialization_encoded", dtype=Int64),
        Field(name="skill_readiness_score", dtype=Float32),
    ],

    source=student_source,

    online=True
)


# -----------------------------
# FEATURE SERVICE
# -----------------------------

student_feature_service = FeatureService(
    name="skill_gap_model_service",
    features=[
        student_feature_view
    ]
)
'''

with open(
    f"{repo_path}/features.py",
    "w"
) as f:
    f.write(feature_definition)
```

```python
# Features = everything except the target
X = training_data[feature_columns].copy()

# Target = Skill Gap
y = training_data["skill_gap"].copy()

print("X columns:")
print(X.columns.tolist())

print("\nTarget values:")
print(y.unique())
```

```python
print(X.dtypes)
```

```python
for col in X.columns:
    if X[col].dtype == "object":
        print(col, X[col].unique())
```

```python
from sklearn.preprocessing import LabelEncoder

X = X.copy()

for col in X.select_dtypes(include="object").columns:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))

print(X.dtypes)
```

```python
print(X.head())
```

```python
y = y.astype(int)

print(y.head())
print(y.unique())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

model.fit(
    X_train,
    y_train
)

print("Model trained successfully! 🎉")
```

```python
predictions = model.predict(X_test)

print(predictions)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test,
    predictions
)

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)
```
